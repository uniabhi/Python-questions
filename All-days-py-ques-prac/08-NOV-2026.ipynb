## Sort Dictionary by value (Without SORT-> Bubble Sort)

```python
def sort_dict_by_value(data):
    """
    Sort a dictionary by its values in ascending order using bubble sort.

    The dictionary is converted into a list of (key, value) pairs, which are
    sorted by value with bubble sort. The sorted pairs are then converted
    back into a dictionary. The original dictionary is not modified.

    Args:
        data (dict): The dictionary to sort. All values must be comparable
            with each other (for example, all numbers or all strings).

    Returns:
        dict: A new dictionary with the same key-value pairs, ordered by
            value from smallest to largest. Pairs with equal values keep
            their original relative order.

    Raises:
        TypeError: If `data` is not a dictionary, or if its values cannot
            be compared with each other (for example, a mix of int and str).

    Side Effects:
        Prints the list of items before and after sorting.

    Example:
        >>> sort_dict_by_value({"apple": 5, "banana": 2, "cherry": 8, "date": 3})
        {'banana': 2, 'date': 3, 'apple': 5, 'cherry': 8}

    Note:
        Bubble sort has O(n^2) time complexity, so it is slow for large
        dictionaries. Dictionary order is preserved only in Python 3.7+.
    """
    if not isinstance(data, dict):
        raise TypeError("input must be a dictionary")
    items = list(data.items())
    # print(items)
    #bubble sort
    for i in range(len(items)):
        for j in range(0, len(items) -i-1):
            if items[j][1] > items[j+1][1]:
                items[j], items[j+1] = items[j+1], items[j]
    # print(items)
    return dict(items)

#main execution
if __name__ == "__main__":
    try:
        data = {"apple": 5, "banana": 2, "cherry": 8, "date": 3}
        print(sort_dict_by_value(data))

    except TypeError as error:
        print("error", error)
    except Exception as e:
        print("unknow error", e)
```

## Find LCM of Two Numbers

```python
def lcm_finder(num1: int, num2: int) -> int:
    """
    Return the LCM (least common multiple) of two integers.

    The LCM is calculated using the formula
    LCM(a, b) = |a * b| / GCD(a, b), where the GCD (greatest common
    divisor) is found with the Euclidean algorithm.

    Args:
        num1 (int): The first integer.
        num2 (int): The second integer.

    Returns:
        int: The LCM of num1 and num2. The result is always non-negative.
            If only one of the numbers is zero, the result is 0.

    Raises:
        TypeError: If either num1 or num2 is not an integer.
        ValueError: If both num1 and num2 are zero, because the LCM is
            not defined in that case.

    Example:
        >>> lcm_finder(9, 18)
        18
        >>> lcm_finder(4, 6)
        12
        >>> lcm_finder(-4, 6)
        12
        >>> lcm_finder(0, 5)
        0
    """
    if not isinstance(num1, int) or not isinstance(num2, int):
        raise TypeError("Both input should be integer")
    if num1 == 0 and num2 == 0:
        raise ValueError("LCM not defined by zeroes")

    def gcd(a, b):
        """Return the greatest common divisor of a and b (Euclidean algorithm)."""
        while b != 0:
            a, b = b, a % b
        return abs(a)

    return abs(num1 * num2) // gcd(num1, num2)


# main execution
if __name__ == "__main__":
    try:
        num1 = 9
        num2 = 18
        print(lcm_finder(num1, num2))
    except (TypeError, ValueError) as error:
        print("error", error)
    except Exception as e:
        print("Unknow error", e)
```

## Convert Binary to Decimal

```python
def decimal_to_binary(binary_str):
    """
    Convert a binary string to its decimal (base-10) integer value.

    The string is read from right to left. Each digit "1" adds
    2 raised to the power of its position (starting at 0 for the
    rightmost digit) to the total.

    Args:
        binary_str (str): A string made of the characters "0" and "1",
            for example "1011".

    Returns:
        int: The decimal value of the binary string. An empty string
            returns 0.

    Note:
        The function does not validate its input. Any character other
        than "1" is treated like "0", so "102" returns 4 instead of
        raising an error.

    Example:
        >>> binary_to_decimal("1011")
        11
        >>> binary_to_decimal("1")
        1
        >>> binary_to_decimal("100000")
        32
        >>> binary_to_decimal("0")
        0
    """
    total = 0
    power = 0
    for digit in reversed(binary_str):
        if digit == "1":
            total = total + 2 ** power
        power = power + 1
    return total
# main execution
if __name__ == "__main__":
    try:
        binary_str = "1011"  #2**5
        print(decimal_to_binary(binary_str))
    except Exception as e:
        print("Unknow error", e)
```

## Swap Two Variables in Python Without Temporary Variable

```python
def swap_two_numbers(a,b):
    a, b = b,a
    return a,b
if __name__ == "__main__":
    try:
        a = 5
        b = 10
        print("a before swap: ", a)
        print("b before swap: ", b)
        result = swap_two_numbers(a,b)
        print(f"a and b after swapping: {result}")
    except Exception as e:
        print(e)
```
