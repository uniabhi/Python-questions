## Nested dectionary question - Python

```python
accounts = {
    "acc1": {"name": "Riya", "Age": 45, "balance" : 85000},
    "acc2": {"name": "Karan", "Age": 29, "balance": 40000},
    "acc3": {"name": "Neha", "Age": 50, "balance": 120000}
}
print(accounts)
```

```python
count = 0
total_acc_bal = 0

for details in accounts.values():
    print(details)
    if details["Age"] > 40:
        total_acc_bal = total_acc_bal + details["balance"]
        count +=1
print(total_acc_bal, ":", count)
```

```python
count = 0
total_acc_bal = 0

accounts = {
    "acc1": {"name": "Riya", "Age": 45, "balance" : 85000},
    "acc2": {"name": "Karan", "Age": 29, "balance": 40000},
    "acc3": {"name": "Neha", "Age": 50, "balance": 120000}
}
for info in accounts.values():
    if info["Age"] > 40:
        print(info.items())
        total_acc_bal = total_acc_bal + info["balance"]
        count+=1
print(total_acc_bal,":", count)
```

```python
def nested_dict(accounts: dict) -> dict:
    """
    """
    count = 0
    total_acc_bal = 0
    for data in accounts.values():
        if data["Age"] > 40:
            total_acc_bal = total_acc_bal + data["balance"]
            count+=1
    return total_acc_bal, count

if __name__ == "__main__":
    try:
        accounts = {
        "acc1": {"name": "Riya", "Age": 45, "balance" : 85000},
        "acc2": {"name": "Karan", "Age": 29, "balance": 40000},
        "acc3": {"name": "Neha", "Age": 50, "balance": 120000}
    }
        result = nested_dict(accounts)
        print(f"The total count of person above age 40 is: {result[1]}")
        print(f"The total balance of {result[1]} person is: {result[0]}")
    except Exception as e:
        print("error:", e)
```

## Sum of Two Integers in Python | Without + Operator

```python
a = 15
b = 1

while b!=0:
    carry =  a & b
    a =  a ^ b
    b =  carry <<  1
print(a)
```

```python
def find__sum_without_plusoperator(a:int, b:int) ->int:
    """
    """
    while b!=0:
        carry = a & b
        a =  a ^ b
        b =  carry<<1
    return a

if __name__ == "__main__":
    try:
        a =  15
        b = 1
        result = find__sum_without_plusoperator(a, b)
        print(f"The sum is: {result}")
    except Exception as e:
        print("Other error:", e)
```
