## Python program to check anagram

```python
def is_anargram(str1:str, str2:str) -> bool:
    """
    check both strings are anagram or not using character counting
    parameters:
        string1(str): first input string
        string(2): second input string
    Returns:
        bool: True if both stings are anagram otherwise False.
    Raises:
        TypeError: if any of the input is not a string.
    """
    if not isinstance(str1, str) or not isinstance(str2, str):
        raise TypeError("both input must be a string")
    str1 = str1.replace(" ", "").lower()
    str2 = str2.replace(" ", "").lower()
    if len(str1) != len(str2):
        return False
    if len(str1) and len(str2) < 2:
        raise ValueError("in both inputs there must atleast 2 characters")
    char_count = {}
    for char in str1:
        if char in char_count:
            char_count[char] += 1
        else:
            char_count[char] = 1
    for char in str2:
        if char not in char_count:
            return False
        char_count[char] -= 1
        if char_count[char] < 0:
            return False
    else:
        return True
if __name__ == "__main__":
    str1 = input("Enter string 1 text:")
    str2 = input("Enter string 2 text:")
    try:
        result = is_anargram(str1, str2)
        if result:
            print("Both strings are anagram")
        else:
            print("Both strings are not anagram")
    except TypeError as error:
        print("error", error)
    except Exception as e:
        print("someother type of error: ", e)
```

```python
def fibonacci_iterative(n: int) -> None:
    """
    print fibonacci sequence using iterative approach

    Fibonacci sequence:
        Each number is the sum of the previous two numbers.
        Example 0 1 1 2 3 5 8 13 21

        Args:
            n(int): number of terms to be print
        Raises:
            ValueError: if n is less than or equal to 0
    """
    if not isinstance(n, int):
        raise TypeError("input must be an integer")
    if n < 0:
        raise ValueError("input should be greater than zero")
    a = 0
    b = 1
    list1 = []
    for i in range(n):
        print(a, end = " ")
        a, b = b, a+b
        list1.append(a)
    print()
    print("your list is", list1)
        
##main execution
if __name__ == "__main__":
    n = int(input("Enter the number n: "))
    try:
        fibonacci_iterative(n)
    except (TypeError, ValueError) as error:
        print("error", error)
    except Exception as e:
        print("someother non-specified error", e)
```

## Python Program to Check Leap Year

```python
def is_leap_year(year:int) -> bool:
    """
    check whether a given year is a leap year or not.
    1. Divisible by 4, 400 and not divisible by 100.
    Args:
        year(int): year to be check if it's a leap or not leap year
    Returns:
        bool: True if leap year otherwise False.
    Raise:
        TypeError: if given input is not integer.
        ValueError: if the year is less than 1582.
    """
    ##edge cases
    if not isinstance(year, int):
        raise TypeError("input must be a integer")
    if year < 1582:
        raise ValueError("input year must be greater than 1582")
    ##main logic
    if year >= 1582 and (year % 4 == 0 and year % 100 !=0) or (year % 400 == 0):
        return True
    else:
        return False
##main execution
if __name__ == "__main__":
    input_year = int(input("Enter the year: "))
    try:
        if is_leap_year(input_year):
            print(f"The given year: {input_year} is a leap year")
        else:
            print(f" {input_year} is not leap year")
    except (TypeError, ValueError) as error:
        print("error", error)
    except Exception as e:
        print("someother non-specified error", e)
```

```python
print(is_leap_year.__doc__)
```

## Python Program to Calculate Power of a Number Without Using ** Operator | Python Interview Question.

```python
def calculate_power(base_value: int, power_value:int) -> int:
    """
    calculate the power of a number without using power (**) operator.

    Args:
        base_value(int): base number.
        Power_value(int): power(must be non-negative).
    Returns:
        int: base_value raised to the power_value.
    Raises:
        TypeError: if either base_value or power_value is not an integer.
        ValueError: if power value is negative.
    """
    ##edge cases
    if not isinstance(base_value, int) or not isinstance(power_value, int):
        raise TypeError("both input must be a integer")
    if power_value < 0:
        raise ValueError("power_value must be a positive number")
    total = 1
    for i in range(power_value):
        total *= base_value
    return total
##main execution
if __name__ == "__main__":
    base_value = int(input("Enter the base value: "))
    power_value = int(input("Enter the power value: "))
    try:
        result = calculate_power(base_value, power_value)
        print(f"The value for base: {base_value} and power: {power_value} = {result}")
    except (TypeError, ValueError) as error:
        print("error", error)
    except Exception as e:
        print("someother non-specified error", e) 
```
