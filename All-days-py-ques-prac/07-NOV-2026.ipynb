## Check if a number is a power of 2

```python
def power_of_two(n:int) -> bool:
    if n<=0:
        return False, None
    power = 0
    while n%2 ==0:
        n = n//2
        power = power + 1
    if n == 1:
        return True, power
    else:
        return False, None

if __name__ == "__main__":
    n = 8
    try:
        result = power_of_two(n)
        print(f"{result}")
    except Exception as e:
        print("Error", e)
```

## How to Flatten a Nested List in Python | Python Coding Interview Question #36

```python
def flatten_list(nested_list:list) ->list:
    """
    Write a function to flatten a list that contains nested list.

    Args:
        nested_list(list): list input from which elements will be flattening

    Return:
        result(list): Flattened list
        
    Raises:
        Exception: unknown error
    """
    result = []
    temp_list = list(nested_list)
    while temp_list:
        item = temp_list.pop()
        if isinstance(item, list):
            temp_list.extend(item)
        else:
            result.append(item)
    return result[::-1]
##main execution
if __name__ == "__main__":
    nested_list = [1,2,[3,[4,5,6],7]]
    try:
        print(flatten_list(nested_list))
    except Exception as e:
        print("Error", e)
```

## Python Interview Question | Find Pairs with Given Sum in a List

```python
def find_pairs(list1:list, target: int | float) -> list:
    """
    Find all pairs of elements in a list that sum up to a given target number.

    Each pair is formed from two different positions (i < j) in the list, so an
    element is never paired with itself. If the list has repeated values, each
    valid combination of positions is returned separately.

    Args:
        list1 (list): List of integers or floats in which pairs will be searched.
        target (int | float): The number that each pair must sum up to.

    Returns:
        list: List of tuples, where each tuple is a pair (a, b) with a + b == target.
              Returns an empty list if no such pair exists.

    Raises:
        TypeError: If list1 is not a list.
        TypeError: If target is not an integer or float.
        TypeError: If any element of list1 is not an integer or float.

    Example:
        >>> find_pairs([1, 2, 3, 4, 5, 6], 7)
        [(1, 6), (2, 5), (3, 4)]
    """
    if not isinstance(target, (int, float)):
        raise TypeError("target must be either integer or float")
    if not isinstance(list1, list):
        raise TypeError("first argument must be a list")
    for x in list1:
        if not isinstance(x, (int, float)):
            raise TypeError(f"{x} is invalid all elements must be integer or float")
    pairs = []
    for i in range(len(list1)):
        for j in range(i, len(list1)):
            if list1[i] + list1[j] == target:
                pairs.append((list1[i] , list1[j]))
    return pairs
##main execution
if __name__ =="__main__":
    try:
        list1 = [1,2,3,4,5,6]
        target = 7
        result = find_pairs(list1, target)
        print(f"pairs are: {result}")
    except TypeError as error:
        print("error",error)
    except Exception as e:
        print("other errors", e)
```

## Convert List of Tuples to Dictionary

```python
def list_to_dict(data:list) -> dict:
    """
    Convert a list of key-value pairs into a dictionary.

    Each pair's first item becomes a key, and its second item becomes
    the corresponding value. If a key appears more than once, the last
    value is used.

    Args:
        data: A list of two-item tuples, such as [("a", 1), ("b", 2)].

    Returns:
        A dictionary mapping each key to its value.
    """
    my_dict = {}
    for key, value in data:
        my_dict[key] = value
    return my_dict
if __name__ == "__main__":
    data = [("a",1), ("b",2),("c",3), ("d", 4)]
    try:
        print(list_to_dict(data))
    except Exception as e:
        print("error", e)
```

## How to Implement Queue in Python using List

```python
class Queue:
    def __init__(self):
        self.queue = []
    def is_empty(self):
        return len(self.queue) ==0
    
    def enqueue(self,value):
        self.queue.append(value)
        print(f"{value} is enqueued, queue: {self.queue}")
    def dequeue(self):
        if self.is_empty():
            print("queue is empty, cannot dequeue")
        else:
            removed = self.queue.pop(0)
            print(f"{removed} dequeue, queue: {self.queue}")
    def peek(self):
        if self.is_empty():
            print("queue is empty")
        else:
            print(f"front element: {self.queue[0]}")
q = Queue()
```

## Find the Median of a List in Python

```python
def find_median(num_list: list) -> int:
    """write a function to calculate the median of a list.
    """
    if not isinstance(num_list, list):
        raise TypeError("input must be a list")
    if len(num_list) == 0:
        raise ValueError("list cannot be empty")
    for num in num_list:
        if not isinstance(num,(int,float)):
            raise ValueError("list elements must be int or float")
    num_list.sort()
    list_length = len(num_list)
    mid = list_length // 2
    if list_length % 2 !=0:
        return (num_list[mid])
    else:
        return (num_list[mid] + num_list[mid-1]) / 2
if __name__ == "__main__":
    num_list = [1,2,3,4,5,6]
    try:
        result = find_median(num_list)
        print(f"The median of list is: {result}")
    except (TypeError, ValueError) as error:
        print("Error", error)
    except Exception as e:
        print("other error", e)
```
