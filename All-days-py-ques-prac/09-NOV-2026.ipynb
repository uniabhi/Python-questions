## Find Common Elements in 3 Lists

```python
def find_common_element(list1:list, list2:list, list3:list) -> list:
    """
    Return the elements common to all three lists, without duplicates.

    Args:
        list1 (list): The first list.
        list2 (list): The second list.
        list3 (list): The third list.

    Returns:
        list: The common elements, in the order they first appear in list1.

    Raises:
        TypeError: If any input is not a list.
    """
    if not isinstance(list1,list) or not isinstance(list2,list) or not isinstance(list3,list):
        raise TypeError("all three inputs must be the lists")

    common_element = []
    
    for item in list1:
        if item in list3 and item in list2 and item not in common_element:
            common_element.append(item)
    return common_element

##main execution
if __name__ == "__main__":
    try:
        list1 = [1,2,3,78,4,5,5,2,3]
        list2 = [3,4,5,6,7,4,78]
        list3 = [5,6,7,8,9,4,78]
        result = find_common_element(list1,list2,list3)
        print(f"The list with common elements is: {result}")
    except TypeError as error:
        print("Error", error)
    except Exception as e:
        print("Unknown error: ", e)
```
