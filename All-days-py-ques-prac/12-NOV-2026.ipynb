## How To Remove Nth occurrence of the character from a string

```python
def remove_nth_occurence(string:str, char_to_remove:str, n:int) -> str:
    """
    Remove the nth occurrence of a character from a string.

    The search ignores case, so "a" matches both "a" and "A". The
    remaining characters keep their original capitalisation.

    Args:
        string (str): The text to process.
        char_to_remove (str): A single character to look for.
        n (int): Which occurrence to remove (1 for the first, 2 for the
            second, and so on). Must be 1 or greater.

    Returns:
        str: The string with the nth occurrence removed. If the character
            appears fewer than n times, the string is returned unchanged.

    Raises:
        TypeError: If string or char_to_remove is not a str, or n is not an int.
        ValueError: If char_to_remove is not exactly one character, or n < 1.

    Example:
        >>> remove_nth_occurence("Apples Are good", "a", 2)
        'Apples re good'
        >>> remove_nth_occurence("banana", "a", 2)
        'banna'
        >>> remove_nth_occurence("hello", "l", 5)
        'hello'
    """
    # input validation
    if not isinstance(string, str):
        raise TypeError("string must be a str")
    if not isinstance(char_to_remove, str):
        raise TypeError("char_to_remove must be a str")
    if len(char_to_remove) != 1:
        raise ValueError("char_to_remove must be a single character")
    if not isinstance(n, int):
        raise TypeError("n must be an int")
    if n < 1:
        raise ValueError("n must be 1 or greater")
        
    occurence_count = 0
    result_string = ""
    for char_in_str in string:
        if char_in_str.lower() == char_to_remove.lower():
            occurence_count = occurence_count + 1
            if occurence_count == n:
                continue   # skip this character, so it is removed
        result_string = result_string + char_in_str
    return result_string
##main execution
if __name__ == "__main__":
    try:
        string = "Apples Are good"
        result = remove_nth_occurence(string, "a", 2)
        print(result)
    except (TypeError, ValueError) as error:
        print("Error", error)
    except Exception as e:
        print("unknown error", e)
```

## Find First Non-Repeating Character in Python String

```python
from typing import Optional


def non_repeated_char(text: str) -> Optional[str]:
    """
    Return the first non-repeated character in a string.

    The function counts how many times each character appears, then
    scans the string from left to right and returns the first character
    that appears exactly once. The check is case-sensitive, so "A" and
    "a" are treated as different characters. Spaces and punctuation
    count as characters too.

    Args:
        text (str): The string to search. Must not be empty.

    Returns:
        str or None: The first character that appears only once, or
            None if every character is repeated.

    Raises:
        TypeError: If text is not a string.
        ValueError: If text is an empty string.

    Example:
        >>> non_repeated_char("swiss")
        'w'
        >>> non_repeated_char("stress")
        't'
        >>> non_repeated_char("Aab")
        'A'
        >>> print(non_repeated_char("aabbcc"))
        None
    """
    # input validation
    if not isinstance(text, str):
        raise TypeError("text must be a string")
    if text == "":
        raise ValueError("text must not be empty")

    # count each character
    char_count = {}
    for char in text:
        if char in char_count:
            char_count[char] += 1
        else:
            char_count[char] = 1

    # return the first character with count 1
    for char in text:
        if char_count[char] == 1:
            return char

    return None


## main execution
if __name__ == "__main__":
    try:
        text = input("Enter the string: ")
        result = non_repeated_char(text)
        if result is None:
            print("No non-repeated character found")
        else:
            print(f"First non-repeated character: {result}")
    except (TypeError, ValueError) as error:
        print("Error:", error)
    except Exception as e:
        print("Other error:", e)
```

## How to Check if a String Contains Only Digits in Python

```python
def check_is_all_digits(text: str) -> bool:
    """
    Check whether a string contains only the digits 0 to 9.

    The function looks at each character and returns False as soon as it
    finds one that is not between "0" and "9". Spaces, signs (+ or -),
    decimal points, and letters all count as non-digits. Only the ASCII
    digits 0-9 are accepted, so digits from other scripts are rejected.

    Args:
        text (str): The string to check.

    Returns:
        bool: True if every character is a digit from 0 to 9. False if
            any character is not a digit, or if the string is empty.

    Raises:
        TypeError: If text is not a string.

    Example:
        >>> check_is_all_digits("12345")
        True
        >>> check_is_all_digits("123a")
        False
        >>> check_is_all_digits("-5")
        False
        >>> check_is_all_digits("3.14")
        False
        >>> check_is_all_digits("")
        False
    """
    # input validation
    if not isinstance(text, str):
        raise TypeError("text must be a string")
    if text == "":
        return False

    for char in text:
        if char < "0" or char > "9":
            return False
    return True


## main execution
if __name__ == "__main__":
    try:
        text = "123a"
        print(check_is_all_digits(text))
    except TypeError as error:
        print("Error:", error)
    except Exception as e:
        print("Unknown error:", e)
```
