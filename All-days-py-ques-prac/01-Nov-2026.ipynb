## Python Program to Find Sum of Digits of a Number

```python
n = 234
total = 0
while n > 0:
    print("Number :", n)
    total = total + n % 10
    print("digit: ", n%10)
    n = n //10
    print("Number after change :", n)
print(total)
```

```python
def sum_of_digits(n: int) -> int:
    """
    Calculate the sum of allpositive integer numbers.
    
    Parameters:
        n(int): summing upon which number will be performed.
    
    Return:
        int: sum of the digits

    Raise:
      TypeError: If the input is not an integer.
      ValueError: If the inputs are not positive.
    """
    ##edge cases
    if not isinstance(n, int):
        raise TypeError("Inputs must be positive integer")
    if n < 0:
        raise ValueError("input digits must be positive number")
    num = 234
    total = 0
    while n > 0:
        digit = n % 10
        total = total + digit
        n = n // 10
    return total
##main excution
if __name__ == "__main__":
    num = input("Enter the positive integer digits: ")

    try:
        num = int(num)
    except ValueError:
        print("Error: ", "Please enter positive integers only. ")
    else:
        try:
            result = sum_of_digits(num)
            print(f" sum of digits {num} is: {result}")
        except (TypeError, ValueError) as error:
            print("error", error)
```

```python
from typing import List, Any 

def remove_duplicate_list(List: List[any]) -> List[any]:
    """
    Removing the duplicates elements from the list while preserving the original order

    Parameters:
        List:List[Any]: A list containing any type of elements.

    Return: 
    List[Any]: will return elemnts in a list without duplicates.

    Raises:
        TypeError: input must be a list of elements.
        ValueError: List cannot be empty.
    """
    ##edge cases: check if input is a list or not
    if not isinstance(List, list):
        raise ValueError("input must be a list.")
    if len(List) == 0:
        raise ValueError("List cannot be empty.")
    if len(List) == 1:
        print("there is only one element is in the List so no duplicate opertation. ")
    return List
    result = []
    for i in List:
        if i not in result:
            result.append(i)
        return result

##main execution

if __name__ == "__main__":
    number_List = [100, 2,2,4,5,6,7,7]
    try:
        result = remove_duplicate_list(number_List)
        print("The elements in the list after duplicate removal is: ", result)
    except (TypeError, ValueError) as error:
        print("error", error)
    except Exception as e:
        print("Something is wrong: ", e)
    else:
        print("After removing duplicates: ", result)
```

## Factorial Using Recursion in Python #11

```python
def factorial(n):
    if n==0 or n==1:
        return 1
    return n * factorial(n-1)
```

```python
def my_factorial(number: int) -> int:
    """
    Calculate the factorial of the given number

    Parameter:
        number:int: the positive input integer no for hwich have to cal. factorial

    Return:
        int: the calculated final result

    Raise:
        TypeError: input must be a integet
        ValueError: input must be a positive number
    """
    ##edge cases
    if not isinstance(number, int):
        raise TypeError("Input must be a integer.")
    if number < 0:
        raise ValueError("input integer must be non-negative number")
    if number == 0 or number ==1:
        return 1
    smaller_answer = my_factorial(number - 1)
    multiplication = number * smaller_answer
    return multiplication

##main execution
if __name__ == "__main__":
    number = int(input("Enter the number: "))
    try:
        result = my_factorial(number)
        print(f"The factorial of {number} is: {result}")
    except (TypeError, ValueError) as error:
        print("error", error)
    except Exception as e:
        print("something is wrong: ", e)
```

## Count Each Occurrence of Characters in a Sentence

```python
def count_characters(input_string: str) -> dict:
    """
    Function to calculate occurences of each characters in a sentence

    Parameters:
        input_string:str - The string whose total characters will have to be counted.

    Return:
        dictionary: the char ocurences will be provided in a key value pair in which value will be count of each char occurences

    Raise:
        TypeError: If the provided input is not a string
    """
    ##edge cases:
    if not isinstance(input_string, str):
        raise TypeError("Input must be a string")
    char_count = {}
    for char in input_string.lower().replace(" ", ""):
        if char in char_count:
            char_count[char] += 1
        else:
            char_count[char] = 1
    return char_count
##main execution
if __name__ == "__main__":
    input_string = input("Enter the sentence: ")
    try:
        result = count_characters(input_string)
        print(f"The count of the provided sentence is : {result}")
    except TypeError as error:
        print("error", error)
    except Exception as e:
        print("something else problem", e)
```
