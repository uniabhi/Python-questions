## Armstrong Number in Python

```python
def is_armstrong_num(num:int) -> bool:
    """
    Check whether an integer is an Armstrong number.

    An Armstrong number (narcissistic number) equals the sum of its own
    digits, each raised to the power of the number of digits.
    Example: 153 = 1^3 + 5^3 + 3^3.

    Args:
        num (int): A non-negative integer to check.

    Returns:
        bool: True if num is an Armstrong number, otherwise False.

    Raises:
        TypeError: If num is not an integer (booleans are rejected too).
        ValueError: If num is negative.

    Examples:
        >>> is_armstrong_num(153)
        True
        >>> is_armstrong_num(9474)
        True
        >>> is_armstrong_num(123)
        False
        >>> is_armstrong_num(0)
        True
    """
    if not isinstance(number, int):
        raise TypeError("input must be a integer")

    if num < 0:
        raise ValueError("input must be a non-negative integer")
    
    digit_count = 0
    temp_num = number
    while temp_num > 0:
        digit_count = digit_count + 1
        temp_num = temp_num //10

    total = 0
    temp_num = number
    while temp_num > 0:
        digit =  temp_num % 10
        total = total + digit ** digit_count
        temp_num = temp_num // 10
    return (number == total)

if __name__ == "__main__":
    try:
        number = int(input("Enter the number:"))
        result = (is_armstrong_num(num))
        print("Is the number is armstrong or not:", result)
    except Exception as error:
        print("error", error)
    except Exception as e:
        print("unknown error", e)
```

## Reverse an Array Without Extra Space | Python Interview Question | O(1)

```python
def reverse_array(nums: list) -> list:
    """
    Reverse a list in place using the two-pointer technique.

    A left pointer starts at the beginning and a right pointer starts at
    the end. At each step the two elements are swapped, then the pointers
    move toward the center until they meet.

    Args:
        nums (list): The list to reverse. An empty list is valid.

    Returns:
        list: The same list object (not a copy), in reversed order.

    Raises:
        TypeError: If nums is not a list.

    Time Complexity:  O(n)
    Space Complexity: O(1)

    Examples:
        >>> reverse_array([1, 2, 3, 4])
        [4, 3, 2, 1]
        >>> reverse_array([])
        []
        >>> reverse_array([7])
        [7]
    """
    # Input validation: only lists are accepted
    if not isinstance(nums, list):
        raise TypeError("input must be a list")

    left = 0
    right = len(nums) - 1

    while left < right:
        nums[left], nums[right] = nums[right], nums[left]
        left += 1
        right -= 1

    return nums


if __name__ == "__main__":
    try:
        nums = [0, 1, 2, 3, 4, 5, 6, 7, 8]
        print(reverse_array(nums))      # [8, 7, 6, 5, 4, 3, 2, 1, 0]
    except TypeError as e:
        print("Error:", e)
    except Exception as e:
        print("other error:", e)
```

## Method - 2 with extra input validation

```python
def total_balance_above_40(accounts: dict) -> tuple:
    """
    Calculate the total balance and number of account holders older than 40.

    Args:
        accounts: A dictionary mapping account IDs to detail dictionaries.
            Each detail dictionary must contain:
            - "age": a non-negative integer
            - "balance": an integer or float

    Returns:
        A tuple (total, count) for account holders whose age is greater than 40.

    Raises:
        TypeError: If accounts or account details are not dictionaries, or if
            age or balance has an invalid type.
        KeyError: If an account is missing "age" or "balance".
        ValueError: If an age is negative.
    """
    if not isinstance(accounts, dict):
        raise TypeError("accounts must be a dictionary")

    total = 0
    count = 0

    for account_id, details in accounts.items():
        if not isinstance(details, dict):
            raise TypeError(
                f"Details for account {account_id!r} must be a dictionary"
            )

        if "age" not in details:
            raise KeyError(f"Account {account_id!r} is missing 'age'")
        if "balance" not in details:
            raise KeyError(f"Account {account_id!r} is missing 'balance'")

        age = details["age"]
        balance = details["balance"]

        # bool is excluded even though Python treats True/False as integers.
        if isinstance(age, bool) or not isinstance(age, int):
            raise TypeError(f"Age for account {account_id!r} must be an integer")
        if age < 0:
            raise ValueError(f"Age for account {account_id!r} cannot be negative")

        if isinstance(balance, bool) or not isinstance(balance, (int, float)):
            raise TypeError(
                f"Balance for account {account_id!r} must be a number"
            )

        if age > 40:
            total += balance
            count += 1

    return total, count


if __name__ == "__main__":
    accounts = {
        "acc1": {"name": "Riya", "age": 50, "balance": 85000},
        "acc2": {"name": "Karan", "age": 60, "balance": 40000},
        "acc3": {"name": "Neha", "age": 38, "balance": 120000},
    }

    try:
        total, count = total_balance_above_40(accounts)
        print(f"Total balance of persons over 40: {total}")
        print(f"Number of persons over 40: {count}")
    except (TypeError, KeyError, ValueError) as error:
        print("Error:", error)
```

## Find the First Duplicate Element in a List | Python Interview Question #52

## Method - 1 without set just with a empty dictionary()

```python
def first_duplicate(numbers: list) -> int | None:
    """
    Find and return the first integer that appears for the second time.

    The function scans the list from left to right and counts each number.
    It returns as soon as a number's count reaches 2.

    Args:
        numbers: A list containing only integers.

    Returns:
        The first repeated integer, or None if there are no duplicates.

    Raises:
        TypeError: If numbers is not a list or contains a non-integer value.
    """
    if not isinstance(numbers, list):
        raise TypeError("Input must be a list")

    char_dict = {}

    for num in numbers:
        if isinstance(num, bool) or not isinstance(num, int):
            raise TypeError("All items in the list must be integers")

        if num not in char_dict:
            char_dict[num] = 1
        else:
            char_dict[num] += 1
            if char_dict[num] == 2:
                return num

    return None


numbers = [10, 5, 3, 4, 3, 5, 6]
print(first_duplicate(numbers))  # 3
```

## Method - 2 with empty set()

```python
def check_duplicate_ele(numbers:list) -> int:
    """
    Return the first integer that appears more than once in a list.

    Args:
        numbers: A list containing only integers.

    Returns:
        The first repeated integer, or None if there are no duplicates.

    Raises:
        TypeError: If numbers is not a list or contains a non-integer value.
    """
    if not isinstance(numbers, list):
        raise TypeError("Input must be a list")
    seen = set()
    for number in numbers:
        if number in seen:
            return number
            break
        else:
            seen.add(number)

if __name__ == "__main__":
    try:
        numbers = [5,4,3,2,6,3,4,5]
        print(check_duplicate_ele(numbers))
    except TypeError as error:
        print("error", error)
    except Exception as e:
        print("other error", e)
```
