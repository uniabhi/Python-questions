## Python Program to Find the Missing Number in a List

```python
def find_missing_number(sample_list:list) -> int:
    """
    Find the first missing number in a list of consecutive integers using adjacent difference comparison.

    Parameters:
        numbers(list): List of integers

    Return:
        int: The first missing number.

    Raises:
        TypeError: If input is not a list or contains non-integers.
        ValueError: If the list has fewer than 2 elements.
        TypeError: If no missing number i found.
    """
    #validate Input
    if not isinstance(sample_list,list):
        raise TypeError("Input must be a list")
    if len(sample_list)<2:
        raise ValueError("List atleast should have 2 elements")
    if not all(isinstance (item, int) for item in sample_list):
        raise TypeError("All list elements must be integers")

    sample_list = sorted(sample_list)
    for i in range(0, len(sample_list)-1):
        if sample_list[i+1] - sample_list[i] !=1:
            return sample_list[i]+1
            break
    else:
        return None
##main execution
if __name__ == "__main__":
    sample_list = [1,2,3,4,5,7,9]
    try:
        print(find_missing_number(sample_list))
    except (TypeError, ValueError) as error:
        print("Error", error)
    except Exception as e:
        print("unknown error",e)                     
```

## Python Program to Find Sum of Squares of N Natural Numbers | Python Interview Question | Q31

```python
total = 0   ##normal way without time complexity
num = 5
for i in range(0, num+1):
    print(i)
    total = total + i*i
print(total)
```

```python
def sum_of_squares(n:int) -> int:
    """
    Return the sum of squares from 1 to n (numbers).
    Uses the formula: n(n+1) * (2*(n+1)) //6

    Parameters:
     n: till that consecutive number have to find sum of squares.

    Return: 
        int: resultant sum (from n no of squares).

    Raises:
        TypeError: if number is not an integer.
        ValueError: if number is not positive.
    """
    ##validate input
    if not isinstance(n, int):
        raise TypeError("input must be an integer type")
    if n < 1:
        raise ValueError("input should be positive")

    total = n*(n+1) * (2*n+1)//6
    return total
##main execution
if __name__ =="__main__":
    num = int(input("Enter the number: "))
    try:
        print(sum_of_squares(num))
    except (TypeError, ValueError) as error:
        print("Error",error)
    except Exception as e:
        print("Unknow error",e)
```

```python
def reverse_integer(num: int) -> int:
    """
    Reverse the digits of an integer, keeping its sign.

    Parameters:
        num (int): The integer to reverse (positive, negative or zero).

    Returns:
        int: The reversed integer. Example: 1234 -> 4321, -1234 -> -4321.

    Raises:
        TypeError: If the input is not an integer.

    Time Complexity:  O(d), where d = number of digits in num.
                      The while loop removes one digit per iteration,
                      so it runs d times. (d is about log10(n), so this
                      is also written as O(log n).)

    Space Complexity: O(1)
                      Only a few variables (sign, reversed_num, digit)
                      are used. No list or extra structure that grows
                      with the input.
    """
    # Validate input (bool is a subclass of int, so reject it explicitly)
    if not isinstance(num, int) or isinstance(num, bool):
        raise TypeError("Input must be an integer")

    sign = -1 if num < 0 else 1   # O(1): remember the sign
    num = abs(num)                # O(1): remove the minus sign
    reversed_num = 0              # O(1): result holder

    while num > 0:                # runs d times -> O(d)
        digit = num % 10                          # take last digit
        reversed_num = reversed_num * 10 + digit  # append digit to result
        num = num // 10                           # drop last digit

    return sign * reversed_num    # O(1): put the sign back


if __name__ == "__main__":
    num = -1234
    try:
        result = reverse_integer(num)
        print("Reverse number :", result)
    except TypeError as e:
        print("Error", e)
    except Exception as e:
        print("Error", e)
```

## Implement Stack Using List in Python (LIFO Explained)

```python
class Stack:
    def __init__(self):
        self.stack = []
    ##push operation
    def push(self, value):
        self.stack.append(value)
        print(f"{value} pushed to the stack")
    ##pop operation
    def pop(self):
        if self.is_empty():
            print("Stack is empty")
            return None
        return self.stack.pop()
    ##peek operation
    def peek(self):
        if self.is_empty():
            print("stack is empty")
            return None
        return self.stack[-1]
    def is_empty(self):
        return len(self.stack) == 0
    
    def __str__(self):
        return str(self.stack)

##main execution
if __name__ == "__main__":
    stack = Stack()
    stack.push(10)
    stack.push(20)
    stack.push(30)
    print("pop:",stack.pop())
    print("peek:",stack.peek())
    print("pop:",stack.pop())
    print("peek:",stack.peek())
    print("pop:",stack.pop())
    print("peek:",stack.peek())
```

## Find Longest Word in a Sentence | Python Interview Question 34

```python
import string
def longest_word_finder(text:str) -> list:
    """
    Function to find the largest word in a sentence
    """
    if not isinstance(text, str):
        raise TypeError("Input must be a string type")
    text = text.lower()
    pattern = str.maketrans("","", string.punctuation)
    clean_text = text.translate(pattern)
    words_list = clean_text.split()
    print(words_list)
    word_count_dict= {}
    for word in words_list:
        if word not in word_count_dict:
            word_count_dict[word] = len(word)
    max_len = max(word_count_dict.values())
    longest_word = [key for key, value in word_count_dict.items() if value == max_len]
    return longest_word
##main execution
if __name__ == "__main__":
    text = " \nThere was one guy which do programming ? in the vilage whith one ? \ laptop and having one / mouse and key-word"
    try:
        print(longest_word_finder(text))
    except TypeError as error:
        print("Error:", error)
    except Exception as e:
        print("other errors:", e)
```
