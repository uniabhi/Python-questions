## Python Program to Print Prime Numbers Between 100 and 200

```python
import math

def is_prime_number(number: int) -> bool:
    """
    Function to check whether a number is prime or not

    Args:
        number (int): The number to check
    Returns:
        bool: True if number is prime, False otherwise.

    Raises:
        TypeError: if the number is not an integer
        ValueError: If number is less than 2 (since primes are >=2).
    """
    #checking number validate type
    if not isinstance(number, int):
        raise TypeError ("input must be an integer")
        #checking number is atleast 2 or not
    if number < 2:
        raise ValueError ("number should be atleast 2 or more")
    for i in range(2, math.isqrt(number) + 1):
        if number % i == 0:
            return False
            break
        else:
            return True

##main exceution
if __name__ == "__main__":
    try:
        prime_numbers = []
        lower_limit = input("Enter lower limit of number: ")
        upper_limit = input("Enter upper limit of the number: ")
        if not lower_limit.strip().isdigit():
            raise TypeError ("Lower linit must be a valid integer (no decimals or letters).")
        if not upper_limit.strip().isdigit():
            raise TypeError ("Lower linit must be a valid integer (no decimals or letters).")
        lower_limit = int(lower_limit)
        upper_limit = int(upper_limit)
        for i in range(lower_limit, upper_limit):
            if is_prime_number(i):
                print(f" {i} : is prime number")
                prime_numbers.append(i)
        print(prime_numbers)
    except (TypeError, ValueError) as e:
        print(e)
    except Exception as e:
        print(e)
```

## Python Program to Find Factorial of a Number | Python Interview Question #5

```python
number = 5
factorial = 1 
for i in range(1, number + 1):
    factorial  = factorial * i
print(factorial)
```

```python
def factorial_num(num : int) -> int:
    """
    Calculate the factorial of the given number
    parameters:
        num(int) - The number for which calculating the factorial
    Returns:
        Int: Factorial of the given number
    Raises:
        TypeError: If the number is not an integer
        ValueError:
            If the number is negative
    """
    if isinstance(num, bool):
        raise TypeError("Boolean values are no allowed")
    if not isinstance(num, int):
        raise TypeError("Number must be an ineteger")
    if num < 0:
        raise ValueError("Number must be an positive integer number")
    factorial = 1
    for i in range(1, num+1):
        factorial = factorial * i
    return factorial

##main execution
if __name__ == "__main__":
    try:
        number = input("Enter the number for which have to calculate the factorial:")
        # if not number.strip().isdigit():
        #     raise TypeError("Input must be an integer not other data types")
        number = int(number)
        print("Factorial is:", factorial_num(number))
    except ValueError as e:
        print("Input must be a positive integer number")
    except TypeTypeError as e:
        print("error", e)
```

## Python Program to Check Palindrome String

```python
def is_palindrome(text: str) -> bool:
    """
    Check whether the passed number is palindrome or not
    Parameters:
        Text: str - input string
    Returns:
        output will be either true or false (Boolean)
    Raises:
        TypeError: if input is not a string
        ValueError: if the input string is empty
    """
    ##edge cases
    ##1. Input must be a string
    if not isinstance(text, str):
        raise TypeError("Input must be a string")
    ##2. If the string is empty
    if text.strip() == "":
        raise ValueError("Input string cannot be empty put atleast two characters")
    else:
        text = text.strip().lower()
    ##Reverse a string
    reverse_string = ""
    for ch in text:
        reverse_string = ch + reverse_string
    return text == reverse_string
##main execution
if __name__ == "__main__":
    try:
        user_input = input("Enter the number for which want to check palindrome : ")
        if is_palindrome(user_input):
            print(f" '{user_input}' is a palindrome")
        else:
            print(f" '{user_input}' is not a palindrome")
    except (TypeError, ValueError) as e:
        print("error", e)
```

## Python Program to Count Vowels in a String | Python Interview Question #7

```python
def count_vowels(text: str) -> int:
    """
    Calculating the total no of vowels in given input string

    Parameters:
        text(int) - user input string

    Return:
        int: Total count of vowels

    Raise:
        TypeError: inptu must be a string
    """
    ##edge cases
    #1. input string is not checking
    if not isinstance(text, str):
        raise TypeError("Input must be a string")
    vowels = "aeiou"
    count = 0
    for char in text.lower():
        if char in vowels:
            count = count + 1
    return count
##main execution
if __name__ == "__main__":
    try:
        user_input = input("Enter the input string: ")
        result = count_vowels(user_input)
        print(f" The total number of vowels in '{user_input}' is: {result} ")
    except TypeError as e:
        print("error", e)
    except Exception as e:
        print(e)
```

```python
from typing import List

def second_largest(nums: list[int]) -> int:
    """
    Find the second largest unique number in a list without using sorting.
    Parameters:
        nums(List[int]): A list of integers.
    Return:
        int: The second largest unique value.
    Raises:
        TypeError: All elements of the list should be integer
        ValueError: If the list has fewer than 2 unique elements.
    """
    #validate the input in a list
    if not isinstance(nums, list):
        raise TypeError("Input must be a list of integers.")
    #validate that all elements in the lists are integers
    if any(not isinstance (current_number, int) for current_number in nums):
        raise TypeError("All elements of the list should be integers")
    #check list has atleast 2 elements
    if len(nums) < 2:
        raise ValueError("List should contain atleast two arguments.")
    if nums[0] > nums[1]:
        largest = nums[0]
        second = nums[1]
    else:
        largest = nums[1]
        second = nums[0]
    for current_number in nums[2:]:
        if current_number > largest:
            second = largest
            largest = current_number
        elif current_number != largest and current_number > second:
            second = current_number
    if second == largest:
        raise ValueError("List must contain atleast two unique element")
    return second
##main execution
if __name__ == "__main__":
    number_list = [1,4,6,8,4, 12,16,17, 19]
    try:
        result = second_largest(number_list)
    except (TypeError, ValueError) as e:
        print("error", e)
    except Exception as e:
        print(e)
    else:
        print("Second largest: " , result)
```
