## Binary Search in Python | Python Program for Binary Search in List | Python Interview Question

```python
def binary_search(sorted_list: list, target:int) -> int:
    """
    Function to perform binary search operation on  a sorted list.
    Args:
        sorted_list(list): A sorted list consist of integers.
        target(int): integer elemnt for which have to search for.
        
        Returns:
            int: index position of the target element if found, if not found than -1.

        Raises:
            TypeError: if the sorted_list is not list and target is not integer.
    """
    if not isinstance(sorted_list, list):
        raise TypeError("sorted list must be a list type.")
    if not isinstance(target, int):
        raise TypeError("target must be a integer")
    low = 0
    high = len(sorted_list) -1
    while low <= high:
        print(high)
        mid = (low+high) //2
        if sorted_list[mid] == target:
            return mid
        elif target < sorted_list[mid]:
            high = mid-1
        else:
            low = mid + 1
    return -1
##main execution
if __name__ == "__main__":
    target = int(input("Enter the single element for target: "))
    sorted_list = [2,5,8,11,13,14,15,17,19,31,32,34,43,54,65,76,87,100]
    try:
        result = binary_search(sorted_list, target)
        print(f"result for given target: {target} is: {result}")
        
    except TypeError as error:
        print("Error", error)
    except Exception as e:
        print("other type of error", e)
```

## Python Program to Reverse Words in a Sentence.

```python
def reverse_word_function(sentence: str) -> str:
    """
    Reverse the order of the words in a sentence.

    Args:
        Sentence(str): input sentence.

    Return:
        str: sentence with words reversed.
    Raises:
        TypeError: If input is not a string.
    """
    ##edge cases
    if not isinstance(sentence, str):
        raise TypeError("input sentence must be a string")

    words_list = []
    word = ""
    for char in sentence:
        if char != " ":
            word = word + char
        else:
            words_list.append(word)
            word = ""
    words_list.append(word)

    rev_sentence = ""
    for i in range(len(words_list)-1, -1,-1):
        rev_sentence = rev_sentence + words_list[i]
        if i != 0:
            rev_sentence = rev_sentence+ " "
    return rev_sentence
##main execution
if __name__ == "__main__":
    sentence = input("Input sentence :")
    try:
        result = reverse_word_function(sentence)
        print(f"The result is: {result}")
    except TypeError as error:
        print("error", error)
    except Exception as e:
        print("other type of error", e)
```

## Python Program to Find GCD Using Euclidean Algorithm

```python
def gcd_find(a:int, b:int) -> int:
    """
    Function to calculate GCD of two numbers using Euclidean algorithm.

    Args:
        a(int): first number
        b(int): second number

    Return:
        int: gcd of a &b.

    Raises:
        TypeError: if both a & b is not integer.
        ValueError: If both numbers are zero
    """
    ##edge cases
    if not isinstance(a, int) or not isinstance(b, int):
        raise TypeError("Both input a and b must be integer")
    if a==0 and b==0:
        raise ValueError("GCD is not defined for both values as zero")
    a, b = abs(a), abs(b)
    while b!=0:
        a, b = b, a%b
    return a
#main execution
if __name__ == "__main__":
    a = int(input("Enter for a: "))
    b = int(input("Enter for b: "))
    try:
        result = gcd_find(a,b)
        print(f"The GCD for both {a} and {b} is: {result}")
    except (TypeError, ValueError) as error:
        print("error", error)
    except Exception as e:
        print("some other type of error", e)
```
