## Reverse a string

```python
import time

input_string = "Python"
rev_string = ""
for char in input_string:
    rev_string = rev_string + char
    new_list = rev_string[::-1]
    time.sleep(1)
print(new_list)
```

```python
def reverse_string(input_string: str) -> str:
    """
    function to reverse a given string
    Args: input_string(str): the string to be reversed
    return: string - reversed version of your input_string
    Raises:
        TypeError: If input is not a string.
    """
# validate data input
    if not isinstance(input_string, str):
        raise TypeError("input must be a string")
    rev_string = ""
    for char in input_string:
        rev_string = char + rev_string
    return rev_string

##main execution file
if __name__ == "__main__":
    try:
        ##Taking input from the user
        user_input = input("Enter your string to reverse")
    ##calling function
        result = reverse_string(user_input)
    #Display the result
        print(f"reverse string: {result}")
    except TypeError as e:
        print(e)
    except Exception as e:
        print(e)
```

```python
def find_largest_number(numbers_list: list[float]) -> float:
    """
    Find the largest number in a list.

    Args:
        numbers_list (list[float]): A list containing int or float values.

    Returns:
        float: The largest number in the list.

    Raises:
        TypeError: If the input is not a list or contains a non-numeric value.
        ValueError: If the list is empty.
    """
    # Validate input type
    if not isinstance(numbers_list, list):
        raise TypeError("Input must be a list")

    # Validate list is not empty
    if len(numbers_list) == 0:
        raise ValueError("List cannot be empty")

    # Validate every element is numeric
    for num in numbers_list:
        if not isinstance(num, (int, float)):
            raise TypeError("All elements must be numeric")

    largest = numbers_list[0]
    for num in numbers_list:
        if num > largest:
            largest = num
    return largest


if __name__ == "__main__":
    try:
        user_input = input("Enter numbers separated by spaces: ")
        numbers_list = [float(item) for item in user_input.split()]
        result = find_largest_number(numbers_list)
        print("Largest number in the list is:", result)
    except (TypeError, ValueError) as error:
        print(error)
    except Exception as e:
        print(e)
```

## Python Program to Check if a Number is Prime

```python
number = 7897328098

for i in range(2, int(num ** 0.5)+1):
    if  number%i ==0:
        print(f"Number: {number} is not prime")
        break
else:
    print(f"Number: {number} is prime")
```

```python
def is_prime_number(number: int) -> bool:
    """
    Function to check whether a number is prim or not

    Args:
        number (int): The number to check
    Returns:
        bool: True if number is prime, False otherwise.

    Raises:
        TypeError: if the number is not an integer
        ValueError: If number is less than 2 (since primes are >=2).
    """
    if not isinstance(number, int):
        raise TypeError ("Input must be an integer value")
    #number should be atleast 2
    if number < 2:
        raise ValueError ("Primes numbers are greater than or equal 2")

    for i in range(2, int(number ** 0.5)+1):
        if number % i ==0:
            return False
            break

    else:
        return True
    #main execution
if __name__ == "__main__":
    try:
        ##taking user input
        user_input  = input("Enter a Number :")
        ##validate and convert input to integer
        if not user_input.strip().isdigit():
            raise TypeError("Input must be a digit")
        number = int(user_input)
        # calling the function
        if is_prime_number(number):
            print(f"Number: {number} is a prime number")
        else:
            print(f"Number: {number} is not prime number")
    except (TypeError, ValueError) as e:
        print(e)
    except Exception as e:
        print("something went wrong")
```
