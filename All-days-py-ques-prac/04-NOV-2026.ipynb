## Python Program to Convert Decimal to Binary Without Using bin()

```python
def decimal_to_binary(decimal: int) -> str:
    """
    Convert decimal number to its corresponding binary representation.

    Args:
        decimal(int): Decimal number.

    Returns:
        str: Binary representation.

    Raises:
        TypeError: If decimal input is not an integer
    """
#edge cases
    if not isinstance(decimal, int):
        raise TypeError("inout must be an integer")
    if decimal==0:
        return "0"
    binary = ""
    while decimal > 0:
        remainder = decimal % 2
        binary = str(remainder) + binary
        decimal = decimal // 2
    return binary
#main execution
if __name__ == "__main__":
    decimal = int(input("Enter the input integer: "))
    try:
        result = decimal_to_binary(decimal)
        print(f"The binary representation of {decimal} is: {result}")
    except TypeError as error:
        print("TypeERROR" , error)
    except Exception as e:
        print("some other error", e)
```

## Intersection of Two Lists Without Using Set in Python Interview

```python
def find_intersection(list1:list, list2:list) -> list:
    """
    Find common elements in both list.

    Args:
        List1(list): Input first.
        List2(list): Input second.

    Return:
        List: output list.

    Raises:
         TypeError: if both inputs are not list.
         ValueError: If inputs are empty.
    """
##edge cases
    if not isinstance(list1, list)  and not isinstance(list2, list):
        raise TypeError("Both inputs must be a list")
    if not list1 or not list2:
        raise ValueError("Both inputs must not be empty")
    result = []
    for item in list1:
        if item in list2 and item not in result:
            result.append(item)
    return result
##edge cases
if __name__ =="__main__":
    list1 = [1,2,100,4,2,5,200]
    list2 = [1,2,300,5,2,500]
    try:
        result = find_intersection(list1, list2)
        print(f"The result is: {result}")
    except TypeError as typeerror:
        print("Typeerror", typeerror)
    except ValueError as valueerror:
        print("value error", valueerror)
    except Exception as e:
        print("other errors", e)
```

## Python Program to Remove Whitespace from a String

```python
text = "   \n       This   \t    is \r      python    "

result = ""

# Remember whether previous character was whitespace
prev_space = False

for char in text.strip():

    # If current character is NOT whitespace
    if not char.isspace():

        # Add normal character to result
        result = result + char

        # Previous character was not whitespace
        prev_space = False

    else:

        # Add a space only if previous character wasn't whitespace
        if prev_space == False:

            result = result + " "

            # Remember that we added a space
            prev_space = True

print(result)
```

```python
def normalize_whitespaces(text:str) -> str:
    """
    Normalizes whitespaces in a string by:
    - Removing leading and trailing whitespaces.
    - Replacing multiple spaces, tabs, newlines with a single space.

    Args:
        text(str): Input string containing extra white spaces.
    Returns:
        str: string with normalized whitespace.

    Raises:
        TypeError: If the input is not a string.
    """
    if not isinstance(text, str):
        raise TypeError("Input must be a string")
    
    result = ""

    # Remember whether previous character was whitespace
    prev_space = False
    
    for char in text.strip():
    
        # If current character is NOT whitespace
        if not char.isspace():
    
            # Add normal character to result
            result = result + char
    
            # Previous character was not whitespace
            prev_space = False
    
        else:
    
            # Add a space only if previous character wasn't whitespace
            if prev_space == False:
    
                result = result + " "
    
                # Remember that we added a space
                prev_space = True
    
    return result
##MAIN EXECUTION
if __name__ == "__main__":
    text = "   \n This is   \n   \t python   "
    try:
        result = (normalize_whitespaces(text))
        print(f"The result is: {result}")
    except TypeError as error:
        print("Typeerror", error)
    except Exception as e:
        print("Other unspecified error", e)
```

## Python Program to Find Sum of List Using Recursion

```python
def recursive_sum(list1: list) -> int:
    """
    Calculate the sum of a list1 elements using recursion.

    Args:
        list1(list): List containing numerical value.

    Return:
        int or float: sum of all elements of list1.

    Raises:
        TypeError: If input is not a list or containing non-numeric values.
        ValueError: If the list is empty.
    """
    if not isinstance(list1,list):
        raise TypeError("Input must be a list")
    if len(list1) ==0:
        raise ValueError("List cannot be empty")
    if not isinstance(list1[0], (int, float)):
        raise TypeError("List can contain only numeric value.")
    if len(list1) == 1:
        return list1[0]
    if len(list1) ==0:
        return 0
    return list1[0]+ recursive_sum(list1[1:])
##main execution
if __name__ == "__main__":
    list1 = [1,2,2,30]
    try:
        print(recursive_sum(list1))
    except (TypeError, ValueError) as error:
        print("error", error)
    except Exception as e:
        print("Unknown error",e)
```
