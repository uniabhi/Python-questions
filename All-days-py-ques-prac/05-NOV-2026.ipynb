## Function to sort a list using buuble sort method.

```python
def sort_list(my_list: list) -> list:
    """
    Sort a list of integers in ascending order using the bubble sort algorithm.

    Parameters:
        my_list(list): A list containing only inetegers.

    Returns:
        List: A list contaning sorted elements in ascending order.

    Raises:
        TypeError:
        - If the input is not a list.
        - If any elements in the list is not an integer.
    """
##edge cases
    if not isinstance(my_list, list):
        raise TypeError("Input is not a list of integers")
    for item in my_list:
        if not isinstance(item, int):
            raise TypeError("All elements of the list should be integers")

    length = len(my_list)
    for i in range(length):
        for j in range(0, length - i-1):
            if my_list[j] > my_list[j+1]:
                my_list[j], my_list[j+1] = my_list[j+1], my_list[j]
    return my_list
#main execution
if __name__ == "__main__":
    my_list = [96, 43, 56, 45, 98, 65]
    try:
        print(sort_list(my_list))
    except TypeError as error:
        print("error", error)
    except Exception as e:
        print("Unknown error", e)
```

## Python Program to Merge Two Sorted Lists

```python
def merge_sorted_list(list1: list, list2:list) -> list:
    """
    Merge two sorted lists of numbers into one sorted list.

    Compares the front elements of both lists, always takes the smaller one,
    and appends it to the result until one list is exhausted. The remaining
    elements of the other list are then appended as they are.

    Args:
        list1 (list): First sorted list containing numeric values.
        list2 (list): Second sorted list containing numeric values.

    Returns:
        list: A new sorted list containing all elements of list1 and list2.

    Raises:
        TypeError: If either input is not a list, or if any element is not
            an int or float.
    """
    if not isinstance(list1, list) or not isinstance(list2, list):
        raise TypeError("Both inputs must be a list")
    for item in list1 + list2:
        if not isinstance(item, (int,float)):
            raise TypeError("List must contain only numbers")
    merged_list = []
    i=j=0
    while i<len(list1) and j<len(list2):
        if list1[i] <= list2[j]:
            merged_list.append(list1[i])
            i= i+1
        else:
            merged_list.append(list2[j])
            j = j+1
    while i < len(list1):
        merged_list.append(list1[i])
        i = i+1
    while j < len(list2):
        merged_list.append(list2[j])
        j = j+1
    return merged_list
##main execution
if __name__ == "__main__":
    list1 = [1,2,3,4]
    list2 = [5,6,7,10]
    try:
        print(merge_sorted_list(list1, list2))
    except TypeError as error:
        print("Error", error)
    except Exception as e:
        print("Unknown error", e)
```

## Python Program to Check if a List is Sorted

```python
def is_sorted(list1:list) -> bool:
    """
    Check whether a list of integers is sorted in ascending order.

    Compares each element with the one after it. If any element is greater
    than the next one, the list is not sorted and the function returns False
    immediately. If the loop finishes without finding such a pair, the list
    is sorted.

    Args:
        list1 (list): List of integer values to check.

    Returns:
        bool: True if the list is sorted in ascending order (or is empty or
        has one element), otherwise False.

    Raises:
        TypeError: If the input is not a list, or if any element is not an
            integer.
    """
##edge case
    if not isinstance(list1, list):
        raise TypeError("Input must be a list")
    for i in range(0, len(list1) -1):
        if not isinstance(list1[i], int) or not isinstance(list1[i+1], int):
            raise TypeError("List elements must be integer")
        if list1[i]> list1[i+1]:
            return False
    return True
##main execution
if __name__ =="__main__":
    list1 = [1,2,3,4]
    try:
        print(is_sorted(list1))
    except TypeError as te:
        print("te", te)
    except Exception as e:
        print("unknown error",e)
```

## Python Program to Count Word Frequency in a Text File

```python
import string
def freq_count():
    """
    Count how often each word appears in a text file.

    Reads the file "clean.txt", removes all punctuation, converts the text to
    lowercase, splits it into words on whitespace, and counts each word using
    a dictionary. The counts are printed as "word: count", one per line.

    The file path is hard-coded to clean.txt inside the
    All-days-py-ques-prac folder.

    Returns:
        None: The function prints results instead of returning them.
            Note that the main block prints this return value, so "None"
            is printed after the counts.

    Raises:
        FileNotFoundError: If clean.txt does not exist at the given path.
        PermissionError: If the file cannot be read.
        UnicodeDecodeError: If the file is not valid text in the default
            encoding.
    """
    with open(r"C:\Users\uniab\MLOPS\YOUTUBE-CODERSINDIA-PY-PRAC-2026-OCT\All-days-py-ques-prac\clean.txt", mode= "r") as file:
        text = file.read()
    pattern = str.maketrans("", "", string.punctuation)
    clean_text = text.translate(pattern)
    words_list = clean_text.lower().split()

    words_count_dict = {}
    for word in words_list:
        if word not in words_count_dict:
            words_count_dict[word] = 1
        else:
            words_count_dict[word] = words_count_dict[word] + 1

    for key, value in words_count_dict.items():
        print(f"{key}: {value}")
#main execution
if __name__ == "__main__":
    try:
        print(freq_count())
    except Exception as e:
        print("Error", e)
```
